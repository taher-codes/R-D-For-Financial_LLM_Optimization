```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
```

```python
model_name = "Qwen/Qwen2.5-1.5B-Instruct"
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
%pip install accelerate
```

```python
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype = "auto", 
    device_map = "auto"
)
```

```python
print(model.device)
```

```python
text = "The company reported record profits and raised its revenue guidance."

messages = [
    {
        "role": "system",
        "content": "You are a financial sentiment classifier. Classify the text as positive, negative, or neutral."
    },
    {
        "role": "user",
        "content": text
    }
]
```

```python
prompt = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)

print(prompt)
```

```python
inputs = tokenizer(
    prompt,
    return_tensors="pt"
).to(model.device)

print(inputs["input_ids"].shape)
```

```python
generated_ids = model.generate(
    **inputs,
    max_new_tokens=20
)
```

```python
new_tokens = generated_ids[0][inputs["input_ids"].shape[1]:]

response = tokenizer.decode(
    new_tokens,
    skip_special_tokens=True
)

print(response)
```
